```python
import pandas as pd
import numpy as np
```

```python
# File paths

train_path = "../data/raw/train_FD001.txt"
test_path = "../data/raw/test_FD001.txt"
rul_path = "../data/raw/RUL_FD001.txt"

train_output_path = "../data/processed/train_FD001_labeled.csv"
test_output_path = "../data/processed/test_FD001_labeled_final_rows.csv"

print("Train path:", train_path)
print("Test path:", test_path)
print("RUL path:", rul_path)
```

```python
# Column names

index_columns = ["unit_number", "time_in_cycles"]

operational_setting_columns = [
    "operational_setting_1",
    "operational_setting_2",
    "operational_setting_3"
]

sensor_columns = [f"sensor_{i}" for i in range(1, 22)]

columns = index_columns + operational_setting_columns + sensor_columns

print("Total columns:", len(columns))
columns
```

```python
# Load raw data files

train_df = pd.read_csv(
    train_path,
    sep=r"\s+",
    header=None,
    names=columns
)

test_df = pd.read_csv(
    test_path,
    sep=r"\s+",
    header=None,
    names=columns
)

rul_df = pd.read_csv(
    rul_path,
    sep=r"\s+",
    header=None,
    names=["true_rul"]
)

print("Training data shape:", train_df.shape)
print("Test data shape:", test_df.shape)
print("RUL labels shape:", rul_df.shape)

display(train_df.head())
display(test_df.head())
display(rul_df.head())
```

```python
print("Training engines:", train_df["unit_number"].nunique())
print("Test engines:", test_df["unit_number"].nunique())

print("\nTraining cycle range:")
display(train_df.groupby("unit_number")["time_in_cycles"].max().describe())

print("\nTest cycle range:")
display(test_df.groupby("unit_number")["time_in_cycles"].max().describe())
```

```python
max_cycle_per_engine = (
    train_df
    .groupby("unit_number")["time_in_cycles"]
    .max()
    .reset_index()
    .rename(columns={"time_in_cycles": "max_cycle"})
)

train_labeled = train_df.merge(max_cycle_per_engine, on="unit_number", how="left")

train_labeled["RUL"] = train_labeled["max_cycle"] - train_labeled["time_in_cycles"]

display(train_labeled.head())
display(train_labeled[["unit_number", "time_in_cycles", "max_cycle", "RUL"]].tail())
```

```python
def assign_risk_label(rul):
    if rul <= 30:
        return "High Risk"
    elif rul <= 75:
        return "Medium Risk"
    else:
        return "Low Risk"

train_labeled["risk_label"] = train_labeled["RUL"].apply(assign_risk_label)

train_labeled["high_risk_flag"] = np.where(train_labeled["RUL"] <= 30, 1, 0)

display(train_labeled[["unit_number", "time_in_cycles", "RUL", "risk_label", "high_risk_flag"]].head(20))

print("Risk label counts:")
display(train_labeled["risk_label"].value_counts())

print("High-risk flag counts:")
display(train_labeled["high_risk_flag"].value_counts())
```

```python
test_max_cycle = (
    test_df
    .groupby("unit_number")["time_in_cycles"]
    .max()
    .reset_index()
    .rename(columns={"time_in_cycles": "max_test_cycle"})
)

rul_df["unit_number"] = rul_df.index + 1

test_labels = test_max_cycle.merge(rul_df, on="unit_number", how="left")

display(test_labels.head())
```

```python
test_final_rows = test_df.merge(test_max_cycle, on="unit_number", how="left")

test_final_rows = test_final_rows[
    test_final_rows["time_in_cycles"] == test_final_rows["max_test_cycle"]
].copy()

test_labeled = test_final_rows.merge(
    rul_df,
    on="unit_number",
    how="left"
)

test_labeled = test_labeled.rename(columns={"true_rul": "RUL"})

test_labeled["risk_label"] = test_labeled["RUL"].apply(assign_risk_label)
test_labeled["high_risk_flag"] = np.where(test_labeled["RUL"] <= 30, 1, 0)

display(test_labeled.head())

print("Test labeled shape:", test_labeled.shape)

print("Test risk label counts:")
display(test_labeled["risk_label"].value_counts())
```
